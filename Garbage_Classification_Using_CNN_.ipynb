```python
CATEGORIES=['Cardboard','Glass','Metal','Paper','Plastic','Trash']
```

```python
import os
import cv2

training = []

def createTrainingData():

    for category_name, path_full in data_paths.items():

        class_num = CATEGORIES.index(category_name)

        if not os.path.exists(path_full):
            print(f"Folder not found: {path_full}")
            continue

        count = 0

        for img_name in os.listdir(path_full):

            img_path = os.path.join(path_full, img_name)

            # Read image
            img_array = cv2.imread(img_path)

            if img_array is None:
                print(f"Skipped: {img_name}")
                continue

            try:
                # Resize image
                new_array = cv2.resize(img_array, (128,128))

                # Store image and label
                training.append([new_array, class_num])

                count += 1

            except Exception as e:
                print(f"Error processing {img_name}: {e}")

        print(f"{category_name}: {count} images loaded")

createTrainingData()

print("Total Images Loaded:", len(training))
```

# Assigning labels and features

* **Features:** Image pixel values are assigned as the input features (**X**) for the CNN model.
* **Labels:** Each image is assigned a corresponding class label (**Y**) based on its garbage category.
* **Data Organization:** Features and labels are stored in separate arrays for efficient processing.
* **Model Input:** The CNN uses features as input and labels as the target output during training.
* **Purpose:** Proper assignment of features and labels enables the model to learn the relationship between images and their correct classes.


```python
X =[]
y =[]
for features, label in training:
  X.append(features)
  y.append(label)
```

# Normalising X and OneHotEncoding on converted labels 0,1,2

* **Normalization** scales image pixel values from **0–255** to **0–1**.
* It improves **training speed** and **model convergence**.
* It reduces the effect of large pixel values, making learning more stable.
* **One-Hot Encoding** converts class labels (0, 1, 2) into binary vectors.
* It enables the CNN to perform **multi-class classification** using **Softmax** and **Categorical Cross-Entropy** loss.


```python
from tensorflow.keras.utils import to_categorical
X = X.astype('float32')
X /= 255

Y = to_categorical(y, len(CATEGORIES))

# to_categorical is a function used to convert categorical class labels (e.g., integers like 0, 1, 2) into one-hot encoded vectors.
# For example, the class 0 would be converted to [1, 0, 0], class 1 to [0, 1, 0], etc.
# This is necessary for classification problems where each class is represented by a vector in the output layer of a neural network.

print(Y[100]) # printing 100th element in Y
print(Y.shape)
```

# Creating the CNN Model

* **Convolutional Layers:** Extract important features such as edges, textures, and shapes from the input images.
* **Max Pooling Layers:** Reduce the size of feature maps while preserving the most important information.
* **Dropout Layers:** Randomly deactivate neurons during training to reduce overfitting and improve generalization.
* **Dense (Fully Connected) Layers:** Combine the extracted features to learn complex patterns and perform classification.
* **Output Layer:** Uses the **Softmax** activation function to predict the probability of each garbage class and identify the final class label.


```python
from keras.models import Sequential
from tensorflow.keras.regularizers import l2
from keras.layers import Conv2D , MaxPooling2D , Flatten , Dense , Dropout

model = Sequential ()  #This initializes an empty model. You will add layers to this model one by one.

#creating 3 convolution layers
model.add(Conv2D(filters=8 , kernel_size=2 , padding="same" ,
                 activation='relu' , input_shape=(128,128,3)))

# using 8 filters in 1st convolution layer, so the output will have 8 feature maps. Fewer filters mean fewer features are detected,
# and thus fewer parameters are learned.
# kernal_size is 2x2
# padding='same' means that the output feature map will have the same width and height as the input.
#input shape is the shape of input image. we had resized it to (224,224,3)

model.add(MaxPooling2D(pool_size=2))

# Pooling is a downsampling operation that reduces the spatial dimensions (height and width) of the input,
#typically by selecting the maximum value from a set of pixels within a small region.
#for each 2x2 region in the input, the max pooling operation will select the maximum value from that region, effectively reducing the size of the feature map.

model.add(Conv2D(filters=16,kernel_size=2,padding='same',activation='relu'))
model.add(MaxPooling2D(pool_size=2))

model.add(Conv2D(filters=32, kernel_size=2, padding='same', activation='relu'))
model.add(MaxPooling2D(pool_size=2))

model.add(Dropout(0.2))

model.add(Flatten())

model.add(Dense(50,activation='relu'))

model.add(Dropout(0.1))
model.add(Dense(len(CATEGORIES), activation='softmax'))   # Corrected to len(CATEGORIES) classes

model.summary()
```

# Compiling the CNN MODEL

* **Compiling** prepares the CNN model for the training process.
* **Optimizer (Adam):** Updates the model weights to minimize prediction errors.
* **Loss Function:** `Categorical Crossentropy` measures the difference between predicted and actual class labels.
* **Evaluation Metric:** `Accuracy` is used to monitor the model's classification performance.
* **Purpose:** Configures the model with the optimizer, loss function, and metrics before training begins.


```python
from keras.optimizers import Adam
model.compile(loss='categorical_crossentropy', optimizer=Adam(learning_rate=0.001), metrics=['accuracy'])
```

# Data Augmentation for CNN Training

* **Data Augmentation** generates new training images by applying random transformations to existing images.
* **Common Transformations:** Rotation, zooming, flipping, shifting, and shearing improve image diversity.
* **Reduces Overfitting:** Helps the CNN learn robust features instead of memorizing the training data.
* **Improves Generalization:** Enables the model to perform better on unseen real-world images.
* **Benefit:** Increases the effective size of the training dataset without collecting additional images.


```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np

# Assuming X is your image data (shape: [num_samples, 512, 512, 3])
# and Y is your one-hot encoded labels (shape: [num_samples, 3])

# Step 1: Create the ImageDataGenerator with augmentations
train_datagen = ImageDataGenerator(
    rotation_range=20,        # Random rotations from -20 to 20 degrees
    width_shift_range=0.2,    # Random horizontal shifts
    height_shift_range=0.2,   # Random vertical shifts
    shear_range=0.2,          # Random shearing transformation(tilting the image across the axis)
    zoom_range=0.2,           # Random zoom
    horizontal_flip=True,     # Random horizontal flip(mirror image)
    fill_mode='nearest' ,  # How to fill pixels after transformations

)


# Step 3: Fit the generators on the data
train_datagen.fit(X_train)
```

* Instead of loading all images into RAM, read them batch by batch. USING ImageDataGenerator

```python
train_generator = train_datagen.flow(X_train, Y_train, batch_size=32)

from tensorflow.keras.callbacks import EarlyStopping
#early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# Step 5: Train the model using the generators
model.fit(
    train_generator,
    steps_per_epoch=len(X_train) // 32,
    epochs=30,
    validation_data=(X_val, Y_val), # Directly pass validation data
    #callbacks=[early_stopping],
    verbose=1
)
```
